```python
#python program  to convert celsuis to fahrenheit (T(%F)T(%)x9/5 +32)

celsius = float(input("Enter temparature in celsius: "))
fahrenheit = celsius * 9/5 + 32
print("Temperature in fahrenheit:", round(fahrenheit,))
```

```python
# python program to convert kilometeres to miles (1 kilometer is equal to 0.62137 miles.)

kilometers = float(input("Enter distance in kilometers: "))
miles = kilometers * 0.62137
print("Distance in miles:", round(miles,))
```
