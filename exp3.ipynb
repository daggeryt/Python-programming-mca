```python
#1.write a python program to change a given string to a new string where the first and last char have been exchanged.[eg:python->nythop]

s=input("Enter a string:")
new_string=s[-1]+s[1:-1]+s[0]
print("New string:",new_string)

```

```python
#2.Write a python program to get a string made of the first 2 and last 2 chars from a given stirng?

s=input("Enter a string:")
new_string=s[:2]+s[-2:]
print("New string:",new_string)
```

```python
#3.Write a pyhton program to create a single string separated with space from two strings by swapping the character at position1,[eg:s1=hello,s2=world,after swap=wello horld?

s1=input("Enter first string:")
s2=input("Enter second string:")
new_s1=s2[0]+s1[1:]
new_s2=s1[0]+s2[1:]
print(new_s1,new_s2)
```

```python
#4. Write a python program to get a string from a given string where all occurance of its first char have been changed to $, expect the first char itself.(use replace()) [ex:onion->oni$n]

s=input("Enter a string:")
new_string=s[0]+s[1:].replace(s[0],"$")
print("New string:",new_string)
```
